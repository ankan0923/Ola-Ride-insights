```python
import pandas as pd
import numpy as np
```

```python
# Read the CSV file into a DataFrame
df= pd.read_csv(r"C:\Users\ankan\Desktop\LabmentixPR\Ola Ride insights\OLA_DataSet.xlsx - July.csv")
df
```

```python
df.shape
```

```python
# 2. Fix column name and data types
df = df.rename(columns={"Vehicle Images": "Vehicle_Images"})
df["Date"] = pd.to_datetime(df["Date"])
```

```python
# 3. Handle missing values (cancelled rides have no payment, rating, etc.)
df["Canceled_Rides_by_Customer"] = df["Canceled_Rides_by_Customer"].fillna("Not Applicable")
df["Canceled_Rides_by_Driver"] = df["Canceled_Rides_by_Driver"].fillna("Not Applicable")
df["Incomplete_Rides"] = df["Incomplete_Rides"].fillna("Not Applicable")
df["Incomplete_Rides_Reason"] = df["Incomplete_Rides_Reason"].fillna("Not Applicable")
df["Payment_Method"] = df["Payment_Method"].fillna("No Payment")
```

```python
# 4. Distance 0 on rides that did not happen is not real, so make it empty
df.loc[df["Booking_Status"] != "Success", "Ride_Distance"] = None
```

```python
# 5. Remove duplicate bookings
df = df.drop_duplicates(subset="Booking_ID")
```

```python
# 6. Add useful columns for the dashboard
df["Hour"] = df["Date"].dt.hour
df["Day"] = df["Date"].dt.day
df["Weekday"] = df["Date"].dt.day_name()
```

```python
# 7. Fill missing values for TAT and Ride Distance with 0, and Ratings with "No Rating"
df[["V_TAT", "C_TAT", "Ride_Distance"]] = df[["V_TAT", "C_TAT", "Ride_Distance"]].fillna(0)
df[["Driver_Ratings", "Customer_Rating"]] = df[["Driver_Ratings", "Customer_Rating"]].fillna("No Rating")
```

```python
# save the cleaned data to a new CSV file
df.to_csv(r"C:\Users\ankan\Desktop\LabmentixPR\Ola Ride insights\ola_cleaned.csv", index=False)
print("Done! Saved ola_cleaned.csv with", len(df), "rows")
```
